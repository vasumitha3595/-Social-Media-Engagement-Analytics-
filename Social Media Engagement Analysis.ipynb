```python
summary = df.groupby(['post_type', 'country', 'sentiment']).agg(
    total_posts=('post_type', 'count'),
    avg_likes=('likes', 'mean'),
    total_shares=('shares', 'sum')
).reset_index()

print(summary)
```

```python
#Statistical Analysis
stat_cols = ["likes", "comments", "shares", "watch_time_sec", "engagement_rate", "follower_count"]
x = df[stat_cols]

summary = pd.DataFrame({

    "mean": x.mean(),
    "median": x.median(),
    "mode": x.mode().iloc[0],
    "std": x.std(),
    "variance": x.var(),
    "min": x.min(),
    "p25": x.quantile(0.25),
    "p75": x.quantile(0.75),
     "max": x.max(),
    "skewness": x.skew(),
    "kurtosis": x.kurt(),
})

summary.round(2)
```

```python
 # Final insight

print("*" * 40)
print("1. POST TYPE PERFORMANCE")
print("=" * 50)

print(
    df.groupby('post_type')
      .agg(
          average_engagement_rate=('engagement_rate', 'mean'),
          average_likes=('likes', 'mean'),
          average_impressions=('impression_count', 'mean')
      )
      .sort_values('average_engagement_rate', ascending=False)
)

print("\n" + "*" * 40)
print("2. CONTENT CATEGORY PERFORMANCE")
print("*" * 40)
print(
    df.groupby('post_category')
      .agg(
          average_engagement_rate=('engagement_rate', 'mean'),
          average_likes=('likes', 'mean'),
          average_impressions=('impression_count', 'mean')
      )
      .sort_values('average_engagement_rate', ascending=False)
)
print("\n" + "*" * 40)
print("3. Country PERFORMANCE")
print("*" * 40)
print(
    df.groupby('country')
      .agg(
          average_engagement_rate=('engagement_rate', 'mean'),
          average_likes=('likes', 'mean'),
          average_impressions=('impression_count', 'mean')
      )
      .sort_values('average_engagement_rate', ascending=False)
)
print("\n" + "*" * 40)
print("4. Verified vs non verified")
print("*" * 40)
print(
    df.groupby('is_verified')
      .agg(
          average_engagement_rate=('engagement_rate', 'mean'),
          average_likes=('likes', 'mean'),
          average_impressions=('impression_count', 'mean'),
          average_watch_time=('watch_time_sec', 'mean')
      )
)

print("\n" + "*" * 40)
print("5. age correlation")
age_corr = df['age'].corr(df['engagement_rate'])

print(f"Correlation between age and engagement rate: {age_corr:.2f}")
print("*" * 40)
print("\n" + "*" * 40)
print("6. TIME OF DAY PERFORMANCE")
print("*" * 40)

print("Time-of-day analysis cannot be performed because")
print("'posted_at' contains dates but not posting times.")


print("\n" + "*" * 40)
print("7. DEVICE PERFORMANCE")
print("*" * 40)

device_performance = (
    df.groupby('device_type')
      .agg(
          average_watch_time=('watch_time_sec', 'mean'),
          average_engagement_rate=('engagement_rate', 'mean'),
          average_impressions=('impression_count', 'mean')
      )
      .sort_values('average_watch_time', ascending=False)
)

print(device_performance)


print("\n" + "*" * 40)
print("8. SENTIMENT PERFORMANCE")
print("*" * 40)

sentiment_performance = (
    df.groupby('sentiment')
      .agg(
          average_engagement_rate=('engagement_rate', 'mean'),
          average_likes=('likes', 'mean'),
          average_comments=('comments', 'mean'),
          average_shares=('shares', 'mean'),
          average_impressions=('impression_count', 'mean')
      )
      .sort_values('average_engagement_rate', ascending=False)
)

print(sentiment_performance)
```
